Eigen Value Eigen Vector

```python
import numpy as np

A = np.array([
    [5, 0, 1],
    [1, 1, 0],
    [-7, 1, 0]
])

values, vectors = np.linalg.eig(A)

values = values.round().astype(int)

for i in range(len(values)):
    v = vectors[:, i]
    v = v / v[2]
    v = v.round(3)

    print(f"Eigenvalue {i+1}: {values[i]}")
    print(f"Eigenvector {i+1}: {v.real}")
```
